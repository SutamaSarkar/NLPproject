```python
from gtts import gTTS
from pydub import AudioSegment

# Your text
tts = gTTS(text=rap_lyrics, lang='en', tld='co.in')
tts.save("voice.mp3")

# Load voice audio
voice = AudioSegment.from_mp3("voice.mp3")

# Load background music (you must upload a music file)
music = AudioSegment.from_mp3("music.mp3")

# Reduce music volume (important!)
music = music - 20

# Loop music if shorter than voice
music = music * (len(voice) // len(music) + 1)

# Trim music to match voice length
music = music[:len(voice)]

# Combine voice + music
final_audio = voice.overlay(music)

# Export final file
final_audio.export("final_rap.mp3", format="mp3")
```
