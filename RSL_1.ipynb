```python
import random
import matplotlib.pyplot as plt
import math
```

```python
#particular N value

N=1000
head_count=0

for i in range(1,N):

    x = random.uniform(0,1)
    if x<=0.5:
        head_count +=1

print("head_count =",head_count)
prob_head=(head_count/N)

print("trial = ",N,", head_count = ",head_count,", prob_head = ",prob_head)
```

```python
trial_list= [5,10,50,100,500,1000,5000,10000,50000,100000,500000,1000000,10000000]
N_list = []
Headcount_list =[]
ProbHead_list =[]

print("TRIALS \t\t HEADCOUNT \t\t PROBOFHEAD")
print("_____________________________________________________________________")

for N in trial_list:

    head_count=0
    for i in range(N):

        x = random.uniform(0,1)
        if x<=0.5:
            head_count +=1

   # print("head_count =",head_count)

    prob_head=(head_count/N)
    print(i,"                    ",head_count,"                   ",prob_head)
   # print(N,head_count,prob_head)

    N_list.append(N)
    Headcount_list.append(head_count)
    ProbHead_list.append(prob_head)

plt.figure(figsize=(6,5))
plt.plot(N_list,ProbHead_list,marker="o",label="Estimated P(H)")
plt.axhline(y=0.5,linestyle="--",color="green",label="Actual P(H)")

plt.xscale("log")
plt.xlabel("Number of Trials(log scale)")
plt.ylabel("P(H)")
plt.title("Number of Trials vs P(H)")
plt.legend()
plt.grid(True)
plt.show()
```
